```python
# surface moyenne et émissions de CO2 selon le type de bâtiment et l'usage principal

stats_usage = df_seattle.groupby(['BuildingType', 'PrimaryPropertyType'])[['PropertyGFABuilding(s)', 'TotalGHGEmissions']].agg(['count', 'mean', 'median'])

# tri du plus gros émetteur de CO2 au plus petit
print(stats_usage.sort_values(('TotalGHGEmissions', 'mean'), ascending=False).round(1))
```

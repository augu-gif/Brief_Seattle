```python
import numpy as np                  # Le moteur de calcul
import pandas as pd                 # Le tableur intelligent
import seaborn as sns               # La visualisation statistique
import matplotlib.pyplot as plt     # La personnalisation graphique
```

```python
PATH = r"C:\Bam\Moi\DataCodes\Seattle\Brief_Seattle\Dataset\2016-building-energy-benchmarking-6af870b7ba821669818300.csv"
df_seattle = pd.read_csv(PATH)
df_seattle.head(11)
```

```python
df_seattle.info()
```

```python
df_seattle.describe()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(data=df_seattle, x="PropertyGFATotal", log_scale=True, kde=True, color="teal")

plt.title("Distribution de la surface totale des bâtiments à Seattle (échelle log)")
plt.xlabel("Surface totale (pieds carrés - échelle logarithmique)")
plt.ylabel("Nombre de bâtiments")

plt.show()
```

```python
table1 = df_seattle.groupby(["PrimaryPropertyType"])["SiteEnergyUse(kBtu)"].mean()
table1 = table1.sort_values(ascending=False).head(5)

plt.bar(x=table1.index, height=table1.values, color="skyblue")

plt.title("Top 5 des types de bâtiments les plus énergivores à Seattle")
plt.xlabel("Consommation moyenne d'énergie (kBtu)")
plt.ylabel("Type de bâtiment")

plt.show()
```

```python
# Selectionner les variables numériques intéressantes
colonnes = ["PropertyGFATotal", "PropertyGFABuilding(s)", "PropertyGFAParking", "SiteEnergyUse(kBtu)", "Electricity(kBtu)", "NaturalGas(kBtu)", "TotalGHGEmissions",]

# Calculer la matrice de corrélation
corr = df_seattle[colonnes].corr()

# Agrandir la figure, afin que les nombres ne se chevauchent pas 
plt.figure(figsize=(10,8))

sns.heatmap(corr, annot = True, fmt=".2f", cmap='coolwarm', vmin=-1, vmax=1)
plt.title("Corrélation entre plusieurs variables numériques")
plt.show()
```
